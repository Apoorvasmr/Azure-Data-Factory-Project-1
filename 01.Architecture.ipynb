# Azure Data Factory Project

## Architecture & Concepts

In this section, we cover the end-to-end data engineering architecture using Azure Data Factory (ADF) and Azure Data Lake Storage (ADLS Gen2) following the Medallion Architecture.

![Azure Data Factory Project Architecture](watermarked_img_8409414530349826317.webp)

## 1. Source Systems (On-Premises & SQL Databases)

* **On-Premises Systems (`ON-PREM`):** 
  * Physical servers, local machines, or legacy data centers hosted locally within an organization rather than on the cloud.
  * Acts as an external data source from which files, logs, or flat data are extracted. 
  * Connected to ADF using a **Self-Hosted Integration Runtime (SHIR)** to securely bridge on-premise data to cloud services.

* **Relational Databases (`SQL`):**
  * Structured relational databases (like SQL Server, PostgreSQL, or MySQL) where transactional data (OLTP) is stored in rows and columns.
  * Serves as a structured data source containing tables (e.g., customer, sales, and product tables) that need to be ingested into Azure.

* **GitHub Repository (`GitHub`):**
  * A cloud-based version control and code hosting platform used to store code, configuration files, or raw source data files (such as sample CSV datasets used during project development).
  * Acts as an external file source from which ADF can pull raw data or CSV files directly via HTTP/REST connections.

## 2. Orchestration & Integration Layer: Azure Data Factory (`ADF`)

* **Simple Definition:** ADF is a cloud-based **ETL / ELT (Extract, Transform, Load)** orchestration service. Think of it as the "factory" or traffic controller that moves and transforms data between different systems.
* **Key Components:**
  * **Linked Services:** Connection strings that securely connect ADF to sources and sinks (e.g., connecting to SQL Database or ADLS Gen2).
  * **Datasets:** Pointers to the actual data structures (tables, files, or folders) inside a linked service.
  * **Pipelines:** Logical groupings of activities (like Copy Activity, Data Flows, and Get Metadata) that execute data workflows.

* **Azure DevOps (`ADo`):**
  * A developer services platform providing CI/CD pipelines, Git repositories, and board tracking.
  * Used for source control integration, enabling collaborative development, branching, pull requests, and automated deployment of ADF pipelines across environments.

## 3. Storage & Medallion Architecture (Azure Data Lake Storage Gen2)

* **Azure Data Lake (`Data Lake`):** 
  * A scalable cloud storage repository that holds vast amounts of raw data in its native format (structured, semi-structured, or unstructured). Enabled with a **Hierarchical Namespace** to support folder structures.
* **Medallion Architecture Layers:**
  1. **Bronze Layer (`🥉` - Raw / Ingestion):** 
     * Stores data exactly as it is received from the source systems with no modifications.
  2. **Silver Layer (`🥈` - Cleansed / Enriched):** 
     * Data is cleaned, filtered, de-duplicated, and structured into optimized file formats (like Parquet or Delta).
  3. **Gold Layer (`🥇` - Curated / Serving):** 
     * Highly refined, aggregated data modeled into **Fact and Dimension tables** (Star Schema) ready for business intelligence and reporting (Power BI).

## 4. Summary of Data Flow
1. **Extract:** Data is pulled from **On-Premises** files and **SQL Databases** using ADF.
2. **Load (Bronze):** ADF lands the raw data into the **Bronze** layer of the Data Lake.
3. **Transform (Silver to Gold):** Data is processed, cleaned into the **Silver** layer, and modeled into business-ready schemas in the **Gold** layer.
