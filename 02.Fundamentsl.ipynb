## Foundational Azure Concepts

Before deploying services like Azure Data Factory or Data Lake Storage, it is essential to set up the foundational management layers in Azure.

### What is a Resource?
A **Resource** is a single, manageable item or service provided by Azure. Every service you provision to build your architecture is a resource. 
*   **Examples:** An Azure Data Factory instance, an Azure Data Lake Storage account, an Azure SQL Database, or an Integration Runtime.

### What is a Resource Group?
A **Resource Group (RG)** is a logical container that holds related resources for an Azure solution. 
*   **Purpose:** It allows you to group resources that share the same lifecycle so you can deploy, manage, update, and delete them together. 
*   **Best Practice:** For this project, all related services (ADF, ADLS Gen2, SQL Database) should be placed inside a single Resource Group to keep billing and management organized.

## Creating a Resource Group

Follow these steps to create a Resource Group for your data engineering project in the Azure Portal:

1. **Log in to Azure:** Navigate to the [Azure Portal](https://portal.azure.com/) and sign in with your credentials.
2. **Locate Resource Groups:** 
   * On the Azure homepage, click on the **Resource groups** icon under "Azure services." 
   * Alternatively, use the top search bar, type *Resource groups*, and select it from the results.
3. **Initiate Creation:** Click the **+ Create** button at the top left corner of the page.
4. **Configure the Basics:**
   * **Subscription:** Select the Azure subscription you are using (e.g., *Free Trial*, *Azure for Students*, or *Pay-As-You-Go*).
   * **Resource group:** Enter a descriptive, unique name (e.g., `rg-adf-project-dev`).
   * **Region:** Select a geographical region close to you or your data sources (e.g., `Central India` or `East US`). This determines where the metadata for your resource group is stored.
5. **Add Tags (Optional but Recommended):** 
   * Click **Next: Tags >**. Tags are name/value pairs used for categorizing resources and tracking billing (e.g., `Project: ADF-Migration`, `Environment: Dev`).
6. **Review and Create:**
   * Click **Review + create** at the bottom of the screen.
   * Azure will run a quick validation. Once you see the green **"Validation passed"** message, click **Create**.
7. **Verify:** Within a few seconds, a notification will appear confirming the creation. Click **Go to resource group** to view your empty container, ready for you to start adding your Data Factory and Data Lake resources.
