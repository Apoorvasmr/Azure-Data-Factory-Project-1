## 🌐 API Ingestion: Extracting Data from GitHub

### What are we doing?
In this step, we will treat GitHub as an external API. We will ingest a JSON file (`DimAirport.json`) directly from a GitHub repository into our Data Lake. To do this, we will use a **Web Activity** to test the API request, and an **HTTP Linked Service** within a **Copy Activity** to actually pull the file.

### ➕ Steps to Configure API Ingestion

#### 1. Get the Raw File Link
* Navigate to your GitHub repository and open the `DimAirport.json` file.
* Click the **Raw** button on the file preview.
* Copy the URL from your browser's address bar. (The base will be `https://raw.githubusercontent.com`).

#### 2. Create the Pipeline & Web Activity
* Create a new pipeline and name it: `API_Ingestion`
* From the **Activities** pane (under *General*), drag a **Web** activity onto the canvas. Name it `weburl`.
* Go to the **Settings** tab of the Web activity:
  * **URL**: Paste the full raw GitHub link you copied in Step 1.
  * **Method**: Select **GET**.

#### 3. Add and Configure the Copy Activity (Source)
* Drag a **Copy data** activity onto the canvas. Name it `copyapidata`.
* Connect the **Web activity** to the **Copy activity** by dragging the green success checkmark box from the Web activity to the Copy activity.
* Click the Copy activity, go to the **Source** tab, and click **+ New**.
* Search for **HTTP** and choose **JSON** as the format.
* **Dataset Name**: `ds_api_source`
* **Linked service**: Click **+ New** to create an HTTP connection:
  * **Name**: `ls_github`
  * **Base URL**: `https://raw.githubusercontent.com`
  * **Authentication type**: **Anonymous** (since the raw GitHub file is public).
  * Click **Create**.
* Back in the dataset connection properties, set the **Relative URL** to the remainder of your specific repository path (e.g., `Apoorvasmr/Azure-Data-Factory-Project-1/refs/heads/main/ADF_project/DimAirport.json`).

#### 4. Configure the Copy Activity (Sink)
* Go to the Copy activity's **Sink** tab and click **+ New**.
* Search for **Azure Data Lake Storage Gen2** and choose **JSON** as the format.
* **Dataset Name**: `ds_api_sink`
* **Linked service**: Select your existing Data Lake linked service (`ls_datalake`).
* **File path**: Specify exactly where it should land in the Data Lake:
  * **Container/Directory/File**: `bronze/github/DimAirport.json`
* **Import schema**: Select **None**.

#### 5. Validate and Execute
* Click **Validate** at the top of the canvas to ensure there are no configuration errors.
* Click **Debug** to run the pipeline.
* ✅ **Result**: The Web activity will make a successful GET request to the GitHub URL, and the Copy activity will pull the JSON data via HTTP and save it as a new file in the `bronze/github/` directory of your Azure Data Lake!
