## 🔵 CI/CD and Version Control: Azure DevOps Integration

In the real world, multiple data engineers collaborate on the same data factory. To do this safely without overwriting each other's work, we integrate ADF with Azure DevOps (ADO) Git. 

### ➕ Step 1: Create the Azure DevOps Project and Repository
*(Note: If you don't have a free ADO account, you can create one quickly by searching for "Azure DevOps" and logging in with your Microsoft account).*

1. Open Azure DevOps and select your organization.
2. Click **New Project**. Name it `ADF project` and click **Create**.
3. Go to the **Repos** tab on the left menu. 
4. Click the repository dropdown at the top center and select **New repository**.
5. **Repository name**: Name it after your ADF instance (e.g., `anshlamba_adf_project`). 
6. Uncheck "Add a README" (keep it completely empty) and click **Create**.

### ➕ Step 2: Configure Git in Azure Data Factory
Now we link our Data Factory to the ADO repo we just made.

1. In ADF, click the **Manage** tab (the toolbox icon on the far left).
2. Under *Source control*, click **Git configuration**, then click **Configure**.
3. **Repository type**: Select **Azure DevOps Git**.
4. **Azure Active Directory**: Choose your Default Directory and click Continue.
5. **Organization name**: Select your ADO organization.
6. **Project name**: Select `ADF project`.
7. **Repository name**: Select your newly created repo.
8. **Collaboration branch**: Select `main`. 
9. Click **Apply**. 

*Notice the top left corner of your ADF screen: You are now working in the `main` branch instead of "Data Factory (Live)".*

### ➕ Step 3: The Developer Workflow (Feature Branches & Pull Requests)
**Industry Rule:** Never make changes directly in the `main` branch! Always create a feature branch.

1. Click the branch dropdown at the top left (currently showing `main`) and select **+ New branch**.
2. **Name**: `feature_ansh` (or your name). Base it on `main`.
3. Let's make a test change: Go to the Author tab, create a new pipeline, and drag a **Wait** activity onto the canvas. 
4. Click **Save** (this acts as a Git Commit, saving the changes *only* to your feature branch).
5. Click the branch dropdown again and select **Create pull request**.
6. This automatically opens Azure DevOps. Click **Create** to submit the PR.
7. Click **Approve**, then click **Complete**. 
8. Use the **Fast-forward merge** method, leave the box checked to delete the feature branch after merging, and click **Complete merge**.

### ➕ Step 4: Publish (Generating ARM Templates)
Now that your feature code is safely merged into `main`, it is time to publish it.

1. Go back to Azure Data Factory and switch your branch back to `main`. (You will now see your merged test pipeline).
2. Click the yellow **Publish** button at the top of the screen.
3. **What happens behind the scenes?** 
   - ADF applies your code to the live Data Factory service.
   - It automatically generates **ARM Templates** (JSON files containing all your ADF configurations).
   - It pushes these templates into a hidden, locked branch called `adf_publish`. 
4. *Your job as a Data Engineer stops here!* A DevOps engineer will later take those ARM templates from the `adf_publish` branch and use Release Pipelines to automatically deploy your work to QA and Production environments.
