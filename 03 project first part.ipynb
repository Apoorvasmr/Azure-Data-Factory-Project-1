## ⚙️ Integration Runtime (IR) Setup

### What is an Integration Runtime?
The Integration Runtime (IR) is the compute infrastructure used by Azure Data Factory to provide data integration capabilities across different network environments. 

### ➕ Creating a New Integration Runtime
To create a new IR, click on **+ New** and select the type of Integration Runtime based on your compute requirements:
* **Azure** (Fully managed cloud compute)
* **Self-Hosted** (For on-premises or private network data)
* **Azure-SSIS** (To execute SSIS packages)
* **Airflow** (Managed Airflow environment)

#### Network Environment Options:
* Azure
* Self-Hosted
* **External Resources**: Linked Self-Hosted

## 🛠️ Self-Hosted Integration Runtime Setup

When you choose to create a **Self-Hosted** Integration Runtime (to connect to on-premises data), follow these configuration steps:

### 1. Settings
* **Name**: Provide a unique name for your Integration Runtime.
* **Self-contained interactive authoring**: Choose to **Enable** or **Disable** this feature based on your security and authoring needs.
* Click **Create**.

### 2. Setup Options
Once created, you will be presented with two ways to install the IR on your local machine or VM:

* **Option 1: Express Setup** (Quick, automated installation directly to your machine)
* **Option 2: Manual Setup** (Preferred for controlled environments)

### 3. Manual Setup Steps
If you choose the Manual Setup, follow this two-step process:

* **Step 1: Download and Install**
  * Click the link to download and install the Integration Runtime. This will open a new window.
  * Choose the specific download version you want and run the installer on your local system.
* **Step 2: Authentication Keys**
  * ADF will generate authentication keys (**Key 1** and **Key 2**). You will need one of these to link your local machine to ADF.

### 4. Local System Configuration
Once the software is installed on your local machine:
1. Search your system for **Microsoft Integration Runtime Configuration Manager** and open it.
2. **Register Integration Runtime (Self-Hosted)**: Copy **Key 1** (or Key 2) from the Azure portal and paste it into the authentication box here.
3. **(Optional) HTTP Proxy**: Configure proxy settings if your corporate network requires it to reach the internet.
4. **Diagnostic Tool**: Use the built-in diagnostic tool to test connections and troubleshoot any network issues.

✅ **Result:** Once successfully registered and running, **now we can use on-prem data** securely in our Azure Data Factory pipelines!

## 🔗 Creating a Linked Service

### What is a Linked Service?
A **Linked Service** in Azure Data Factory (ADF) acts like a connection string. It defines the connection information and authentication credentials needed for ADF to securely connect to external data stores or compute resources. Think of it as giving ADF the "keys and address" to access your data.

---

### ➕ Setting up a File System Linked Service
Now that our compute infrastructure (Integration Runtime) is ready, we need to create a Linked Service to connect ADF to our on-premises local files.

**1. Create New**
* Go to the Linked Services section and click **+ New**.
* Search for and select **File System** (this allows us to connect to a local or network drive).

**2. Configuration Details**
* **Integration Runtime**: Select the **Self-Hosted Integration Runtime** that we created and registered in the previous steps. This tells ADF to use your local machine's compute to make the connection.
* **Host**: Provide the folder or drive path you want to access. 
  * *For this project*: We will use a local path like `C:\` or a specific folder path on your machine.
  * *🏢 Company Context*: In a real-world corporate environment, you wouldn't use a local `C:\` drive. Instead, you would use a secure network file share, a dedicated host server, or an FTP path.

**3. Authentication (Username & Password)**
To allow ADF to read your local files, you must provide your machine's login credentials. You can find or use your username in a few ways:

* **Option 1: Microsoft Account**
  If your Windows PC is linked to your Microsoft account, use your **Microsoft Email** and the corresponding **Password** (or Windows PIN/App password depending on your security setup).
  
* **Option 2: Windows Settings**
  Navigate to your PC's **Settings > Accounts > Your info / View Profile** to find the exact local username registered on your machine.

* **Option 3: Command Prompt (Most Accurate)**
  To get the exact domain and username format that the system expects:
  1. Open **Command Prompt** (cmd).
  2. Type `echo %username%` and press Enter to see your current user.
  3. Alternatively, type `whoami` and press Enter. This will output your username in the `DOMAIN\username` or `HOSTNAME\username` format, which is highly recommended to use in ADF.

**4. Final Verification**
* Enter your chosen Username and Password into the Linked Service configuration.
* Click **Test Connection**. 
* ✅ **Result**: If it succeeds, it means Azure Data Factory has successfully communicated with your local File System via the Self-Hosted Integration Runtime! You are now ready to build pipelines with your local data.

## 🗄️ Creating an Azure Storage Account

### What is an Azure Storage Account?
An **Azure Storage Account** is a secure, scalable, and highly available cloud storage solution provided by Microsoft. It contains all of your Azure Storage data objects: blobs (files), file shares, queues, tables, and disks. 

In an Azure Data Factory (ADF) project, we typically use it as a **Data Lake (ADLS Gen2)** to act as a central repository. It is used to store raw data ingested from on-premises systems, stage data during transformations, and save final processed datasets.

### ➕ Steps to Create a Storage Account

**1. Navigate to Storage Accounts**
* Log in to the Azure Portal.
* In the search bar at the top, search for **Storage accounts** and select it.
* Click on **+ Create**.

**2. Basics Tab**
* **Subscription**: Select your active Azure subscription.
* **Resource Group**: Choose your project's resource group (or create a new one to keep all project-related resources organized together).
* **Storage account name**: Provide a globally unique name (must be 3-24 characters, using only lowercase letters and numbers).
* **Region**: Select the region closest to you or your data sources to minimize latency.
* **Performance**: Choose **Standard** (general-purpose v2) for most common workloads.
* **Redundancy**: Select **Locally-redundant storage (LRS)** for a cost-effective development environment. (Use higher redundancy like GRS for actual production environments).

**3. Advanced Tab (Crucial for Data Projects)**
* Under the **Data Lake Storage Gen2** section, check the box for **Enable hierarchical namespace**. 
  * *Why?* This is a critical step. It upgrades standard blob storage into a true Data Lake, allowing for traditional folder-level directory structures and significantly optimizing performance for big data analytics.

**4. Review and Create**
* You can leave the remaining tabs (Networking, Data protection, Encryption, Tags) as their defaults for a standard setup.
* Click **Review**. 
* Once Azure validates your settings, click **Create**. Wait a few moments for the deployment to finish.

**5. Create a Container (Your Cloud Folder)**
Once the deployment is complete, you need to create a container to actually hold your files:
1. Go to your newly created Storage Account resource.
2. On the left-hand menu under **Data storage**, click on **Containers**.
3. Click **+ Container**, give it a logical name (e.g., `raw-data`, `staging`, or `bronze`), and click **Create**. 

✅ **Result:** Your cloud storage is now fully provisioned! The next step in your project will be creating a new Linked Service in Azure Data Factory (just like you did for the File System) to connect to this Storage Account, allowing you to copy your local data into the cloud.
