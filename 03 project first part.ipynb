## ⚙️ Integration Runtime (IR) Setup

### What is an Integration Runtime?
The Integration Runtime (IR) is the compute infrastructure used by Azure Data Factory to provide data integration capabilities across different network environments. 

### ➕ Creating a New Integration Runtime
To create a new IR, click on **+ New** and select the type of Integration Runtime based on your compute requirements:
* **Azure** (Fully managed cloud compute)
* **Self-Hosted** (For on-premises or private network data)
* **Azure-SSIS** (To execute SSIS packages)
* **Airflow** (Managed Airflow environment)

#### Network Environment Options:
* Azure
* Self-Hosted
* **External Resources**: Linked Self-Hosted

## 🛠️ Self-Hosted Integration Runtime Setup

When you choose to create a **Self-Hosted** Integration Runtime (to connect to on-premises data), follow these configuration steps:

### 1. Settings
* **Name**: Provide a unique name for your Integration Runtime.
* **Self-contained interactive authoring**: Choose to **Enable** or **Disable** this feature based on your security and authoring needs.
* Click **Create**.

### 2. Setup Options
Once created, you will be presented with two ways to install the IR on your local machine or VM:

* **Option 1: Express Setup** (Quick, automated installation directly to your machine)
* **Option 2: Manual Setup** (Preferred for controlled environments)

### 3. Manual Setup Steps
If you choose the Manual Setup, follow this two-step process:

* **Step 1: Download and Install**
  * Click the link to download and install the Integration Runtime. This will open a new window.
  * Choose the specific download version you want and run the installer on your local system.
* **Step 2: Authentication Keys**
  * ADF will generate authentication keys (**Key 1** and **Key 2**). You will need one of these to link your local machine to ADF.

### 4. Local System Configuration
Once the software is installed on your local machine:
1. Search your system for **Microsoft Integration Runtime Configuration Manager** and open it.
2. **Register Integration Runtime (Self-Hosted)**: Copy **Key 1** (or Key 2) from the Azure portal and paste it into the authentication box here.
3. **(Optional) HTTP Proxy**: Configure proxy settings if your corporate network requires it to reach the internet.
4. **Diagnostic Tool**: Use the built-in diagnostic tool to test connections and troubleshoot any network issues.

✅ **Result:** Once successfully registered and running, **now we can use on-prem data** securely in our Azure Data Factory pipelines!
